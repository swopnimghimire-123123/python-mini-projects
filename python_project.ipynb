```python
import time
import random

def chat_game():
    # Jokes list
    jokes = [
        "Why don’t programmers like nature? Too many bugs.",
        "Why did the computer go to the doctor? Because it caught a virus.",
        "Why do Python programmers wear glasses? Because they can’t C.",
        "Why was the math book sad? Because it had too many problems."
    ]

    # Facts list
    facts = [
        "The human brain has around 86 billion neurons.",
        "Python was named after the TV show 'Monty Python’s Flying Circus'.",
        "Honey never spoils. Archaeologists have found edible honey in ancient tombs.",
        "Octopuses have three hearts.",
        "The first computer bug was an actual insect."
    ]

    print("Computer: Hi there!")

    # 1. Ask name
    name = input("Computer: What is your first name? \nYou: ")
    print(f"\nComputer: Nice to meet you, {name}!")

    # 2. Ask how they are doing
    mood = input(f"\nComputer: Hey {name}, how are you doing today?\nYou: ").lower()

    if "good" in mood or "fine" in mood or "great" in mood:
        print("Computer: That’s really nice to hear.")
    elif "bad" in mood or "sad" in mood or "tired" in mood:
        print("Computer: I’m sorry to hear that, I hope things get better.")
    else:
        print("Computer: I see. Thanks for sharing that with me.")

    # 3. Ask what they are working on
    work = input(f"\nComputer: What are you currently working on, {name}?\nYou: ")
    print(f"Computer: Wow! {work} sounds really interesting.")

    # 4. Ask how they are feeling right now
    feeling = input(f"\nComputer: How are you feeling right now?\nYou: ").lower()

    if "happy" in feeling:
        print("Computer: That’s amazing! Keep smiling.")
    elif "sad" in feeling:
        print("Computer: It’s okay to feel sad sometimes. You’re not alone.")
    elif "stressed" in feeling or "anxious" in feeling:
        print("Computer: Take a deep breath, You’re doing your best.")
    else:
        print("Computer: Thanks for telling me. Your feelings matter.")

    # 5. Ask about games
    game_type = input(
        f"\nComputer: Do you like online games or offline games?\nYou: "
    ).lower()

    if "online" in game_type:
        print("Computer: Online games are fun! Playing with others is exciting.")
    elif "offline" in game_type:
        print("Computer: Offline games are awesome too! Peaceful and focused.")
    else:
        print("Computer: Games are fun in any form!")

    # 6. Ask if they want a joke
    joke_choice = input(
        f"\nComputer: Do you want to listen to a joke? (yes/no)\nYou: "
    ).lower()

    if "yes" or "ofc" or "go ahead" or "would love that" in joke_choice:
        print("Computer:", random.choice(jokes))
    else:
        print("Computer: No worries! Maybe next time.")

    # 7. Ask if they want to know the time
    time_choice = input(
        f"\nComputer: Do you want to know what time it is right now? (yes/no)\nYou: "
    ).lower()

    if "yes" or "ofc" or "go ahead" or "would love that" in time_choice:
        current_time = time.strftime("%I:%M %p")
        print(f"Computer: The current time is {current_time}.")
    else:
        print("Computer: Alright! Time can wait.")

    # 8. Another interesting question
    dream = input(
        f"\nComputer: What is one thing you want to achieve in the future?\nYou: "
    )
    print(f"Computer: That’s a wonderful goal, {name}! Keep working towards it.")

    # 9. Ask if they want to hear a fact
    fact_choice = input(
        f"\nComputer: Do you want to listen to an interesting fact? (yes/no)\nYou: "
    ).lower()

    if "yes" or "ofc" or "go ahead" or "would love that" in fact_choice:
        print("Computer:", random.choice(facts))
    else:
        print("Computer: No problem! Facts can wait too.")

    # 10. Final message
    print(f"\nComputer: It was really nice talking to you, {name}.")
    print("Computer: Take care and have a great day!")

# Start the chat game
chat_game()
```
