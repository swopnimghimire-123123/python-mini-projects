```python
"""
Group 5 – Expanded Text-Based Chat Game
----------------------------------------
This program is an enhanced version of a simple text chat game.
The computer talks with the user for 10 turns.
Features:
 - Asks questions
 - Provides multiple choice options
 - Remembers user mood and interests
 - Gives dynamic, conversational responses

Submitted by: Group 5
"""

import random

def ask_choice_question(prompt, options):
    """
    Prints a question with multiple options and returns the chosen text.
    """
    print("\nComputer:", prompt)
    for i, opt in enumerate(options, 1):
        print(f"  {i}. {opt}")

    while True:
        choice = input("Choose an option (1/2/3): ").strip()
        if choice in ["1", "2", "3"]:
            return options[int(choice) - 1]
        print("Invalid input! Please choose 1, 2, or 3.")


def get_followup_based_on_topic(topic):
    """
    Ask deeper follow-up questions based on topic.
    """

    if topic == "games":
        return random.choice([
            "What type of games do you enjoy the most?",
            "Do you like single-player or multiplayer games?",
            "What game have you spent the most hours on?"
        ])

    if topic == "study":
        return random.choice([
            "What subject do you enjoy studying the most?",
            "Are you preparing for exams or learning for fun?",
            "What is the hardest topic you’ve studied recently?"
        ])

    if topic == "life":
        return random.choice([
            "What was the best moment of your week?",
            "Is there something you are excited about right now?",
            "Do you prefer peaceful days or adventurous ones?"
        ])

    if topic == "ai":
        return random.choice([
            "Do you think AI will help people become more productive?",
            "If you built a robot, what would you want it to do?",
            "Do you feel excited or scared about future technology?"
        ])

    return "That's interesting! Tell me more."


def generate_bot_reply(user_text, memory):
    """
    Smarter reply logic with mood + topics + deeper questions
    """

    text = user_text.lower()

    # Mood detection
    moods = {
        "good": "That's awesome! I love hearing positive things.",
        "great": "Amazing! What made your day so good?",
        "happy": "Nice! Happiness is contagious.",
        "sad": "I'm sorry. Want to talk about what made you feel sad?",
        "tired": "Rest is important. Have you been studying a lot?",
        "stressed": "Take a deep breath. What’s stressing you out?"
    }

    for mood in moods:
        if mood in text:
            memory["mood"] = mood
            return moods[mood]

    # Topic detection
    topics = {
        "game": "games",
        "play": "games",
        "study": "study",
        "learning": "study",
        "life": "life",
        "robot": "ai",
        "ai": "ai",
        "future": "ai"
    }

    for key in topics:
        if key in text:
            memory["topic"] = topics[key]
            return get_followup_based_on_topic(memory["topic"])

    # Fallback generic responses
    generic = [
        "Interesting! Tell me more.",
        "Why do you think so?",
        "Hmm… that sounds meaningful.",
        "Oh really? Keep going.",
        "That’s something I’d like to hear more about."
    ]

    return random.choice(generic)


def chat_game():
    """
    Runs the expanded chat game with options, questions, mood memory,
    branching conversations, and 10 interaction turns.
    """

    MAX_TURNS = 10
    turn = 1

    memory = { "mood": None, "topic": None }

    print("Computer: Hello! I’m your chat buddy for today.")
    print("Computer: Let’s talk for 10 turns. Type 'quit' anytime to stop early.\n")

    # First question to break the ice
    choice = ask_choice_question(
        "How are you feeling right now?",
        ["Good 😊", "Okay 😐", "Not so good 😢"]
    )

    if "good" in choice.lower():
        memory["mood"] = "good"
        print("Computer: Nice! What made your day good?")
    elif "okay" in choice.lower():
        memory["mood"] = "neutral"
        print("Computer: I see. Want to talk about something fun?")
    else:
        memory["mood"] = "sad"
        print("Computer: I'm here for you. What's troubling you?")

    while turn <= MAX_TURNS:

        user_text = input(f"\nYou ({turn}/{MAX_TURNS}): ").strip()

        if user_text == "":
            print("Computer: Can you type something? I didn’t get it.")
            continue

        if user_text.lower() == "quit":
            print("\nComputer: Alright, ending chat early. Take care!")
            return

        bot_reply = generate_bot_reply(user_text, memory)
        print("Computer:", bot_reply)

        # After 4 turns, ask a choice-based topic
        if turn == 4:
            topic_choice = ask_choice_question(
                "What do you want to talk about next?",
                ["Games 🎮", "Studies 📘", "Life 🌱"]
            )

            if "games" in topic_choice.lower():
                memory["topic"] = "games"
            elif "studies" in topic_choice.lower():
                memory["topic"] = "study"
            else:
                memory["topic"] = "life"

            print("Computer:", get_followup_based_on_topic(memory["topic"]))

        turn += 1

    print("\nComputer: That was 10 turns! I had fun chatting with you. Game over!")


# Run the game
if __name__ == "__main__":
    chat_game()
```
